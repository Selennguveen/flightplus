```python
# ============================================================
# Sürekli Sayısal Değişkenler — Histogram + KDE
# ============================================================
from scipy.stats import gaussian_kde
from plotly.subplots import make_subplots

surekli_sayisal_cols = ["Age", "Flight Distance", "Departure Delay in Minutes", "Arrival Delay in Minutes"]
hist_colors = ["#0B3D66", "#1E6FB0", "#3E8FC4", "#59A8DE"]  # kart renkleriyle aynı sıra
x_araliklari = [None, None, [0, 250], [0, 250]]  # gecikme sütunlarında uzun kuyruğu kırpıp gövdeyi net gösteriyoruz

fig_dist = make_subplots(rows=2, cols=2, subplot_titles=surekli_sayisal_cols, vertical_spacing=0.14, horizontal_spacing=0.08)

for i, col in enumerate(surekli_sayisal_cols):
    row, col_pos = divmod(i, 2)
    veri = df_train[col].dropna()

    fig_dist.add_trace(
        go.Histogram(
            x=veri,
            histnorm="probability density",
            marker=dict(color=hist_colors[i], line=dict(color=AIRLINE_BG, width=0.5)),
            opacity=0.75,
            name=col,
            showlegend=False,
            hovertemplate=f"<b>{col}</b><br>Değer: %{{x}}<br>Yoğunluk: %{{y:.4f}}<extra></extra>",
        ),
        row=row + 1, col=col_pos + 1,
    )

    kde = gaussian_kde(veri)
    x_kde = np.linspace(veri.min(), veri.max(), 300)
    y_kde = kde(x_kde)

    fig_dist.add_trace(
        go.Scatter(
            x=x_kde, y=y_kde, mode="lines",
            line=dict(color=AIRLINE_INK, width=2.5),
            name="KDE", showlegend=False,
            hoverinfo="skip",
        ),
        row=row + 1, col=col_pos + 1,
    )

    if x_araliklari[i] is not None:
        fig_dist.update_xaxes(range=x_araliklari[i], row=row + 1, col=col_pos + 1)

fig_dist.update_annotations(font=dict(family=AIRLINE_FONT, color=AIRLINE_INK, size=13))

fig_dist.update_layout(
    template="airline",
    title=dict(text="Sürekli Sayısal Değişkenlerin Dağılımı (Histogram + KDE)"),
    height=650,
    margin=dict(t=90, b=50, l=50, r=30),
)

fig_dist.show()
```

```python
# ============================================================
# Çarpıklık (Skewness) ve Basıklık (Kurtosis)
# ============================================================
from IPython.display import display


def skew_yorum(deger):
    if abs(deger) < 0.5:
        return "yaklaşık simetrik"
    elif abs(deger) < 1:
        return "orta derecede çarpık"
    else:
        yon = "sağa" if deger > 0 else "sola"
        return f"yüksek derecede {yon} çarpık"


def kurtosis_yorum(deger):
    if deger > 1:
        return "sivri (leptokurtic) — ağır kuyruklu"
    elif deger < -1:
        return "basık (platykurtic) — hafif kuyruklu"
    else:
        return "normale yakın (mesokurtic)"


skew_kurt_ozet = pd.DataFrame({
    "Skewness": df_train[surekli_sayisal_cols].skew().round(3),
    "Kurtosis": df_train[surekli_sayisal_cols].kurt().round(3),
})
skew_kurt_ozet["Çarpıklık Yorumu"] = skew_kurt_ozet["Skewness"].apply(skew_yorum)
skew_kurt_ozet["Basıklık Yorumu"] = skew_kurt_ozet["Kurtosis"].apply(kurtosis_yorum)

display(skew_kurt_ozet)
```

```python
# ============================================================
# satisfaction × Kategorik Değişkenler — %100 Yığılmış Bar Grafik
# Her kategori için memnun / memnun değil oranı (satır bazlı yüzde)
# ============================================================
kategorik_cols_biv = ["Gender", "Customer Type", "Type of Travel", "Class"]

# satisfaction icin sabit renk kurali: satisfied = teal, neutral-or-dissatisfied = coral
satisfaction_colors = {"satisfied": "#2FA891", "neutral or dissatisfied": "#E8834F"}

fig_biv_cat = make_subplots(
    rows=2, cols=2,
    subplot_titles=kategorik_cols_biv,
    horizontal_spacing=0.12, vertical_spacing=0.16,
)

for i, col in enumerate(kategorik_cols_biv):
    row, col_pos = divmod(i, 2)
    oran_tablo = pd.crosstab(df_train[col], df_train["satisfaction"], normalize="index") * 100
    kategoriler = oran_tablo.index.tolist()

    for satisfaction_deger in ["neutral or dissatisfied", "satisfied"]:
        fig_biv_cat.add_trace(
            go.Bar(
                y=kategoriler,
                x=oran_tablo[satisfaction_deger],
                name=satisfaction_deger,
                orientation="h",
                marker=dict(color=satisfaction_colors[satisfaction_deger]),
                showlegend=(i == 0),
                hovertemplate=f"<b>%{{y}}</b><br>{satisfaction_deger}: %{{x:.1f}}%<extra></extra>",
            ),
            row=row + 1, col=col_pos + 1,
        )

fig_biv_cat.update_layout(
    template="airline",
    title=dict(text="satisfaction × Kategorik Değişkenler (Kategori İçi Yüzde Dağılımı)"),
    barmode="stack",
    height=750,
    legend=dict(orientation="h", yanchor="bottom", y=1.06, xanchor="center", x=0.5),
    margin=dict(t=110, b=30, l=110, r=20),
)
fig_biv_cat.update_xaxes(ticksuffix="%", range=[0, 100])
fig_biv_cat.update_annotations(font=dict(family=AIRLINE_FONT, color=AIRLINE_INK, size=12))

fig_biv_cat.show()
gorseli_kaydet(fig_biv_cat, "10_satisfaction_x_kategorik_degiskenler")
```

```python
# ============================================================
# Renklendirilmiş Scatter Plot — Age × Ortalama Hizmet Puani × satisfaction
# İki sürekli değişken + hedef değişkeni tek grafikte görüyoruz
# (100k+ satırı doğrudan çizmek hem yavaş hem dosya boyutunu şişirir,
# bu yüzden stratified bir örneklem alıyoruz — oranlar korunuyor)
# ============================================================
scatter_ornek = df_train.groupby("satisfaction", group_keys=False).apply(
    lambda g: g.sample(n=2500, random_state=42)
).copy()

# Bu grafik feature engineering hücresinden bağımsız çalışabilsin.
if "Ortalama Hizmet Puani" not in scatter_ornek.columns:
    scatter_ornek["Ortalama Hizmet Puani"] = scatter_ornek[service_cols].mean(axis=1)

satisfaction_colors = {"satisfied": "#2FA891", "neutral or dissatisfied": "#E8834F"}

fig_multi_scatter = go.Figure()

for satisfaction_deger, renk in satisfaction_colors.items():
    alt_kume = scatter_ornek[scatter_ornek["satisfaction"] == satisfaction_deger]
    fig_multi_scatter.add_trace(
        go.Scatter(
            x=alt_kume["Age"],
            y=alt_kume["Ortalama Hizmet Puani"],
            mode="markers",
            name=satisfaction_deger,
            marker=dict(color=renk, size=5, opacity=0.45),
            hovertemplate=f"<b>{satisfaction_deger}</b><br>Yaş: %{{x}}<br>Ort. Hizmet Puanı: %{{y:.2f}}<extra></extra>",
        )
    )

fig_multi_scatter.update_layout(
    template="airline",
    title=dict(text="Age × Ortalama Hizmet Puanı (satisfaction'a Göre Renklendirilmiş, 5.000 Örneklem)"),
    xaxis=dict(title="Age"),
    yaxis=dict(title="Ortalama Hizmet Puanı (0-5)"),
    height=560,
    legend=dict(orientation="h", yanchor="bottom", y=1.04, xanchor="center", x=0.5),
    margin=dict(t=80, b=50, l=60, r=30),
)

fig_multi_scatter.show()
gorseli_kaydet(fig_multi_scatter, "15_age_x_ortalama_hizmet_puani_scatter")
```

```python
# ============================================================
# PCA (Temel Bileşen Analizi) — 18 Sayısal Değişkeni 2 Boyuta İndirme
# Amaç: memnun/memnun olmayan yolcuların 2 boyutta ne kadar ayrıştığını görmek
# ============================================================
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer

pca_cols = surekli_sayisal_cols + service_cols

# PCA NaN kabul etmediği için, hücre sırasından bağımsız olarak eksikleri dolduruyoruz.
pca_imputer = SimpleImputer(strategy="median")
X_pca_imputed = pca_imputer.fit_transform(df_train[pca_cols])

pca_scaler = StandardScaler()
X_pca_scaled = pca_scaler.fit_transform(X_pca_imputed)

pca = PCA(n_components=2, random_state=42)
pca_bilesenler = pca.fit_transform(X_pca_scaled)

pca_df = pd.DataFrame(pca_bilesenler, columns=["PC1", "PC2"], index=df_train.index)
pca_df["satisfaction"] = df_train["satisfaction"].values

# görselleştirme için aynı stratified örneklemi kullanıyoruz (performans için)
pca_ornek = pca_df.groupby("satisfaction", group_keys=False).apply(lambda g: g.sample(n=2500, random_state=42))

fig_pca = go.Figure()

for satisfaction_deger, renk in satisfaction_colors.items():
    alt_kume = pca_ornek[pca_ornek["satisfaction"] == satisfaction_deger]
    fig_pca.add_trace(
        go.Scatter(
            x=alt_kume["PC1"],
            y=alt_kume["PC2"],
            mode="markers",
            name=satisfaction_deger,
            marker=dict(color=renk, size=5, opacity=0.4),
            hovertemplate=f"<b>{satisfaction_deger}</b><br>PC1: %{{x:.2f}}<br>PC2: %{{y:.2f}}<extra></extra>",
        )
    )

fig_pca.update_layout(
    template="airline",
    title=dict(
        text=f"PCA — 2 Boyuta İndirgenmiş Görünüm (Açıklanan Varyans: %{(pca.explained_variance_ratio_.sum()*100):.1f})"
    ),
    xaxis=dict(title=f"PC1 (%{pca.explained_variance_ratio_[0]*100:.1f} varyans)"),
    yaxis=dict(title=f"PC2 (%{pca.explained_variance_ratio_[1]*100:.1f} varyans)"),
    height=560,
    legend=dict(orientation="h", yanchor="bottom", y=1.04, xanchor="center", x=0.5),
    margin=dict(t=80, b=50, l=60, r=30),
)

fig_pca.show()
gorseli_kaydet(fig_pca, "16_pca_2_boyut")

# PC1'e en çok katkı veren değişkenler
pc1_agirliklar = pd.Series(pca.components_[0], index=pca_cols).sort_values(key=abs, ascending=False)
print("PC1'e en çok katkı veren 5 değişken:")
print(pc1_agirliklar.head(5).round(3))
```

Bu, hizmet puanının satisfaction ile güçlü ilişkisini (istatistiksel testlerde de doğruladığımız gibi) görsel olarak teyit ediyor; Age ekseninde ise böyle net bir ayrışma yok.</li>
    <li><b>PCA</b>: 18 sayısal değişkeni 2 boyuta indirdiğimizde, bu iki bileşen toplam varyansın yaklaşık %35'ini açıklıyor — düşük bir oran değil ama tüm bilgiyi de yakalamıyor, veri setinin gerçekten çok boyutlu olduğunu gösteriyor. PC1'e en çok katkı veren değişkenler <b>Inflight entertainment, Cleanliness, Seat comfort, Food and drink, Online boarding</b> — yani PC1 büyük ölçüde "kabin/hizmet deneyimi kalitesi" ekseni.     <li><b>Genel özet</b>: Multivariate analiz, bivariate analizin gösteremediği iki önemli şeyi ortaya çıkardı — değişkenler arası <b>etkileşim etkisi</b> (Class × Type of Travel) ve verinin <b>modele uygunluğuna dair erken bir sinyal</b> (PCA ayrışması).       Bivariate analizde bazı değişkenlerde büyük farklar gördük (örneğin Class × satisfaction),
      bazılarında ise neredeyse hiç fark yoktu (Gender gibi). Ama grafikte "büyük duruyor" demek
      istatistiksel olarak yeterli değil — bu farkın <b>rastlantıyla mı</b> oluştuğunu, yoksa

      <span class="aft-badge">Class</span>) için <b>ki-kare (Chi-square) bağımsızlık testi</b>;
      sürekli sayısal değişkenler ve hizmet puanları için ise <b>Mann-Whitney U testi</b>
      (verimiz normal dağılıma uymadığı için t-testi yerine bunu tercih edeceğiz).
    </p>
    <p>
      Her testte <b>p-değeri</b>'ne bakacağız: p &lt; 0.05 ise "bu değişken ile satisfaction
      arasındaki ilişki istatistiksel olarak anlamlı" diyebileceğiz. Ayrıca sadece anlamlılığı değil,
      ilişkinin <b>ne kadar güçlü</b> olduğunu da (Cramér's V ile) ölçeceğiz — çünkü çok büyük bir
      veri setinde neredeyse her fark "anlamlı" çıkabilir, asıl önemli olan etkinin büyüklüğü.

```python
# ============================================================
# Ki-Kare (Chi-square) Bağımsızlık Testi
# Kategorik değişkenler ile satisfaction arasındaki ilişkinin
# anlamlılığını (p-değeri) ve gücünü (Cramér's V) ölçüyoruz
# ============================================================
from IPython.display import display
from scipy.stats import chi2_contingency

kategorik_cols_test = ["Gender", "Customer Type", "Type of Travel", "Class"]
chi_kare_sonuclari = []

for col in kategorik_cols_test:
    tablo = pd.crosstab(df_train[col], df_train["satisfaction"])
    chi2, p, dof, beklenen = chi2_contingency(tablo)

    n = tablo.values.sum()
    k = min(tablo.shape) - 1
    cramers_v = np.sqrt(chi2 / (n * k))

    chi_kare_sonuclari.append({
        "Değişken": col,
        "Chi-square": round(chi2, 2),
        "p-değeri": p,
        "Anlamlı mı? (p<0.05)": "Evet" if p < 0.05 else "Hayır",
        "Cramér's V": round(cramers_v, 4),
    })

chi_kare_df = pd.DataFrame(chi_kare_sonuclari).sort_values("Cramér's V", ascending=False)
display(chi_kare_df)
```

```python
# ============================================================
# Mann-Whitney U Testi
# Sürekli sayısal değişkenler + hizmet puanları için, satisfaction
# gruplarının dağılımları arasında anlamlı fark var mı test ediyoruz
# (verimiz normal dağılıma uymadığı için t-testi yerine bu tercih edildi)
# ============================================================
from IPython.display import display
from scipy.stats import mannwhitneyu

sayisal_cols_test = surekli_sayisal_cols + service_cols
grup_memnun = df_train.loc[df_train["satisfaction"] == "satisfied"]
grup_memnun_degil = df_train.loc[df_train["satisfaction"] == "neutral or dissatisfied"]

mw_sonuclari = []

for col in sayisal_cols_test:
    a = grup_memnun[col].dropna()
    b = grup_memnun_degil[col].dropna()
    u, p = mannwhitneyu(a, b, alternative="two-sided")

    # rank-biserial korelasyon: etki büyüklüğü (-1 ile +1 arası)
    etki_r = 1 - (2 * u) / (len(a) * len(b))

    mw_sonuclari.append({
        "Değişken": col,
        "p-değeri": p,
        "Anlamlı mı? (p<0.05)": "Evet" if p < 0.05 else "Hayır",
        "Etki Büyüklüğü (r)": round(etki_r, 3),
    })

mw_df = pd.DataFrame(mw_sonuclari)
mw_df["|r|"] = mw_df["Etki Büyüklüğü (r)"].abs()
mw_df = mw_df.sort_values("|r|", ascending=False).drop(columns="|r|").reset_index(drop=True)
display(mw_df)
```

    <li><b>Bu değerler hata değil, gerçek</b>: Uzun bir gecikme (örneğin birkaç saatlik) veri girişi hatası değil, gerçekten yaşanmış bir olay — hava koşulları, teknik arıza gibi nedenlerle. Bu yüzden bu satırları silmek veya değerlerini "düzeltmek" bilgi kaybına yol açar; aksine, uzun gecikmelerin satisfaction üzerindeki etkisini modelin görebilmesi bizim için değerli.</li>
    <li><b>Karar</b>: Aykırı değerleri <b>silmiyoruz ve olduğu gibi bırakıyoruz</b>. Bunun iki gerekçesi var: (1) projede öncelikli olarak kullanacağımız <b>ağaç bazlı modeller (XGBoost, LightGBM)</b> aşırı değerlere karşı doğal olarak dayanıklıdır — bölme (split) mantığıyla çalıştıkları için birkaç uç değerden etkilenmezler; (2) eğer ileride karşılaştırma amaçlı bir <b>Logistic Regression baseline</b> denersek, gecikme sütunlarındaki çarpıklığı gidermek için satırları silmek yerine <b>log dönüşümü (log1p)</b> gibi bir özellik mühendisliği adımını feature engineering aşamasında ayrıca değerlendireceğiz.</li>
    <li>Özetle: aykırı değer <b>tespiti</b> yaptık (EDA'da), ama <b>müdahale</b> etmemeyi seçtik — bu, veriyi olduğu gibi koruyup kararı modele bırakan bilinçli bir tercih.</li>
  </ul>
</div>


```python
# ============================================================
# Log Dönüşümü — Departure/Arrival Delay için Türetilmiş Sütunlar
# Orijinal sütunlar korunuyor (ağaç bazlı modeller için), yanına
# log1p ile dönüştürülmüş halleri ekleniyor (ölçeğe duyarlı
# modeller — örn. Logistic Regression — için ileride kullanılabilir)
# ============================================================
gecikme_cols = ["Departure Delay in Minutes", "Arrival Delay in Minutes"]

for veri_seti, isim in [(df_train, "df_train"), (df_test, "df_test")]:
    for col in gecikme_cols:
        log_col = f"{col} (log)"
        veri_seti[log_col] = np.log1p(veri_seti[col])

    print(f"{isim}: log sütunları eklendi -> " + ", ".join(f"{c} (log)" for c in gecikme_cols))

print()
print("Çarpıklık (skewness) karşılaştırması (df_train):")
for col in gecikme_cols:
    orijinal_skew = df_train[col].skew()
    log_skew = df_train[f"{col} (log)"].skew()
    print(f"  {col:32s}: orijinal={orijinal_skew:.2f}  ->  log1p={log_skew:.2f}")
```

  <div class="afn-head">🧭 Karar — Kategorik Değişken Kodlama (Encoding)</div>
  <ul>
    <li><b>Gender, Customer Type, Type of Travel</b> — üçü de ikili (binary) değişken: sırasıyla (Male/Female), (Loyal Customer/disloyal Customer), (Business travel/Personal Travel). Bunlar için basit <b>0/1 kodlama</b> yeterli — herhangi bir sıralama anlamı taşımadıkları için hangisinin 0 hangisinin 1 olduğu modelin performansını etkilemez.</li>
    <li><b>Class</b> farklı — 3 sınıfı var (Eco, Eco Plus, Business) ve bunlar arasında <b>doğal bir sıralama</b> var: Eco &lt; Eco Plus &lt; Business (hem konfor hem de bivariate analizde gördüğümüz memnuniyet oranı açısından). Bu yüzden rastgele bir One-Hot Encoding yerine, bu sırayı koruyan <b>ordinal (sıralı) kodlama</b> kullanıyoruz: Eco=0, Eco Plus=1, Business=2. Böylece model "Business, Eco Plus'tan daha yüksek bir sınıf" bilgisini doğrudan sayısal ilişki olarak görebiliyor.</li>
    <li>Bu kodlamayı hem `df_train` hem `df_test` üzerinde aynı eşleme (mapping) ile uyguluyoruz — tutarlılık için, `LabelEncoder`'ı ayrı ayrı fit etmek yerine sabit bir sözlük (dictionary) kullanıyoruz; böylece train ve test'te aynı kategori her zaman aynı sayıya karşılık gelir.</li>
  </ul>
</div>


```python
# ============================================================
# Kategorik Değişken Kodlama (Encoding)
# İkili değişkenler için 0/1, Class için sırayı koruyan ordinal kodlama
# ============================================================
from IPython.display import display

ikili_esleme = {
    "Gender": {"Male": 0, "Female": 1},
    "Customer Type": {"Loyal Customer": 0, "disloyal Customer": 1},
    "Type of Travel": {"Personal Travel": 0, "Business travel": 1},
}
class_esleme = {"Eco": 0, "Eco Plus": 1, "Business": 2}

for veri_seti in [df_train, df_test]:
    for col, esleme in ikili_esleme.items():
        veri_seti[f"{col} (encoded)"] = veri_seti[col].map(esleme)

    veri_seti["Class (encoded)"] = veri_seti["Class"].map(class_esleme)

display(df_train[["Gender", "Gender (encoded)", "Customer Type", "Customer Type (encoded)",
                   "Type of Travel", "Type of Travel (encoded)", "Class", "Class (encoded)"]].head())
```

  <div class="afn-head">🧭 Karar — Sayısal Değişken Ölçekleme (Scaling)</div>
  <ul>
    <li>Ağaç bazlı modeller (XGBoost, LightGBM) ölçeklemeye ihtiyaç duymaz, ama henüz hangi modelin en iyi performansı vereceğini bilmiyoruz — karşılaştırma için bir <b>Logistic Regression baseline</b> da deneyeceğiz, ve doğrusal modeller farklı ölçeklerdeki sütunlardan (yaş: 7-85, mesafe: 31-4983, hizmet puanı: 0-5) olumsuz etkilenir. Bu yüzden ölçeklenmiş sütunları da hazırlıyoruz.</li>
    <li><b>Hangi sütunlar?</b> Age, Flight Distance, gecikme sütunlarının log halleri (log dönüşümü yaptığımız için ham hali değil log halini ölçekliyoruz) ve 14 hizmet puanı — encoding yaptığımız kategorik sütunlar zaten 0/1/2 gibi küçük ve sınırlı olduğu için ölçeklemeye gerek yok.</li>
    <li><b>Veri sızıntısını önlemek için önemli kural</b>: `StandardScaler`'ı sadece `df_train` üzerinde <code>fit</code> ediyoruz, sonra hem train hem test'e <code>transform</code> uyguluyoruz. Eğer test setini de fit'e dahil edersek, test setindeki istatistiksel bilgi (ortalama, standart sapma) modele sızmış olur — bu, gerçek dünyada asla olmayacak bir avantaj sağlar ve model performansını yanıltıcı şekilde iyi gösterir.</li>
    <li>Sonuçları orijinal sütunların yanına <code>(scaled)</code> ekiyle yeni sütunlar olarak ekliyoruz — orijinal değerler yine korunuyor, hangi modelin hangi sütun setini kullanacağına modelleme aşamasında karar vereceğiz.</li>
  </ul>

```python
# ============================================================
# Sayısal Değişken Ölçekleme (StandardScaler)
# Sadece df_train üzerinde fit edilip, hem train hem test'e
# transform uygulanıyor (veri sızıntısını önlemek için)
# ============================================================
from IPython.display import display
from sklearn.preprocessing import StandardScaler

olcek_cols = ["Age", "Flight Distance", "Departure Delay in Minutes (log)", "Arrival Delay in Minutes (log)"] + service_cols

scaler = StandardScaler()
scaler.fit(df_train[olcek_cols])

for veri_seti, isim in [(df_train, "df_train"), (df_test, "df_test")]:
    olceklenmis = scaler.transform(veri_seti[olcek_cols])
    olceklenmis_df = pd.DataFrame(
        olceklenmis,
        columns=[f"{c} (scaled)" for c in olcek_cols],
        index=veri_seti.index,
    )
    veri_seti[olceklenmis_df.columns] = olceklenmis_df
    print(f"{isim}: {len(olcek_cols)} sütun ölçeklendi")

display(df_train[["Age", "Age (scaled)", "Flight Distance", "Flight Distance (scaled)"]].head())
```

  <div class="afn-head">🧭 Karar — Yeni Özellik Türetme (Feature Engineering)</div>
  <ul>
    <li><b>Ortalama Hizmet Puanı</b>: 14 hizmet puanının ortalaması, tek bir "genel deneyim" skoru olarak. `satisfaction` ile korelasyonu <b>0,496</b> — tek başına bile oldukça güçlü bir sinyal, çünkü tüm hizmet puanlarının ortak bilgisini tek bir sütunda topluyor.</li>
    <li><b>Toplam Gecikme</b> (Departure + Arrival) ve <b>Gecikme Farkı</b> (Arrival − Departure): gecikmenin toplam büyüklüğünü ve uçuş sırasında büyüyüp büyümediğini yakalıyor. Korelasyonları zayıf (~-0,03 ile -0,06) ama yine de modele ek, yorumlanabilir bir bilgi sağlıyorlar.</li>
    <li><b>Yaş Grubu</b> (18-25, 26-40, 41-60, 60+): bivariate analizde Age ile satisfaction arasında zayıf-orta bir ilişki görmüştük, ama bu <b>doğrusal değil</b> — gruplara ayırınca ortaya çıkıyor: 41-60 yaş grubunda memnuniyet oranı %57,7 ile en yüksek, ama 60+ grubunda %20,8'e düşüyor. Yani "yaş arttıkça memnuniyet artar" gibi basit bir ilişki yok; orta-üst yaş grubu en memnun, hem genç hem çok yaşlı gruplar daha az memnun. Bu tür bir U-şeklindeki (non-linear) ilişkiyi yakalamak için yaşı gruplamak, ham `Age` sütununa göre modele daha fazla bilgi veriyor.</li>
    <li>Bu dört özelliği de hem `df_train` hem `df_test`'e aynı mantıkla ekliyoruz — hiçbiri train'den öğrenilen bir istatistiğe dayanmıyor (aritmetik işlemler), bu yüzden veri sızıntısı riski yok.</li>

```python
# ============================================================
# Yeni Özellik Türetme (Feature Engineering)
# ============================================================
from IPython.display import display

for veri_seti in [df_train, df_test]:
    veri_seti["Ortalama Hizmet Puani"] = veri_seti[service_cols].mean(axis=1)
    veri_seti["Toplam Gecikme"] = veri_seti["Departure Delay in Minutes"] + veri_seti["Arrival Delay in Minutes"]
    veri_seti["Gecikme Farki"] = veri_seti["Arrival Delay in Minutes"] - veri_seti["Departure Delay in Minutes"]
    veri_seti["Yas Grubu"] = pd.cut(
        veri_seti["Age"],
        bins=[0, 25, 40, 60, 100],
        labels=["18-25", "26-40", "41-60", "60+"],
    )

display(df_train[["Age", "Yas Grubu", "Ortalama Hizmet Puani", "Toplam Gecikme", "Gecikme Farki"]].head())
```

```python
# ============================================================
# Model Tipine Göre Özellik Seti Tanımlama
# ============================================================
kategorik_encoded_cols = ["Gender (encoded)", "Customer Type (encoded)", "Type of Travel (encoded)", "Class (encoded)"]
turetilmis_cols = ["Ortalama Hizmet Puani", "Toplam Gecikme", "Gecikme Farki"]
log_scaled_cols = [f"{c} (log) (scaled)" for c in gecikme_cols] + [f"{c} (scaled)" for c in ["Age", "Flight Distance"] + service_cols]

# Yas Grubu doğrusal olmayan bir ilişki gösterdiği için (bkz. cell 78) one-hot encoding ile ekleniyor
yas_grubu_dummies = pd.get_dummies(df_train["Yas Grubu"], prefix="Yas Grubu")

# --- Ağaç bazlı modeller (XGBoost, LightGBM) için özellik seti ---
# Ham sütunlar kullanılabilir; ölçeklemeye gerek yok, çoklu doğrusal bağlantıdan etkilenmezler
agac_ozellik_seti = (
    surekli_sayisal_cols + service_cols + kategorik_encoded_cols + turetilmis_cols
    + yas_grubu_dummies.columns.tolist()
)

# --- Doğrusal modeller (Logistic Regression) için özellik seti ---
# Ölçeklenmiş/log'lu sütunlar tercih edilir; Toplam Gecikme çıkarıldı çünkü
# Departure/Arrival Delay (log) (scaled) ile neredeyse birebir aynı bilgiyi taşıyor
# (mükemmele yakın çoklu doğrusal bağlantı — bkz. cell 80 tartışması)
dogrusal_ozellik_seti = (
    log_scaled_cols + kategorik_encoded_cols + ["Ortalama Hizmet Puani"]
    + yas_grubu_dummies.columns.tolist()
)

print(f"Ağaç bazlı model özellik sayısı: {len(agac_ozellik_seti)}")
print(f"Doğrusal model özellik sayısı: {len(dogrusal_ozellik_seti)}")
```

```python
# ============================================================
# X / y Ayrımı
# df_train_split, df_val ve df_test için özellik (X) ve hedef (y) matrislerini oluşturuyoruz
# ============================================================
def ozellik_matrisi_olustur(df, ozellik_seti):
    """Yas Grubu için one-hot encoding uygular ve verilen özellik setine göre X matrisini döner."""
    df_dummies = pd.get_dummies(df["Yas Grubu"], prefix="Yas Grubu")
    df_genis = pd.concat([df, df_dummies], axis=1)
    # sütun sırası/tutarlılığı için eksik olan dummy sütunları 0 ile dolduruyoruz
    return df_genis.reindex(columns=ozellik_seti, fill_value=0)


# Ağaç bazlı modeller için X/y
X_train_agac = ozellik_matrisi_olustur(df_train_split, agac_ozellik_seti)
X_val_agac = ozellik_matrisi_olustur(df_val, agac_ozellik_seti)
X_test_agac = ozellik_matrisi_olustur(df_test, agac_ozellik_seti)

# Doğrusal modeller için X/y
X_train_dogrusal = ozellik_matrisi_olustur(df_train_split, dogrusal_ozellik_seti)
X_val_dogrusal = ozellik_matrisi_olustur(df_val, dogrusal_ozellik_seti)
X_test_dogrusal = ozellik_matrisi_olustur(df_test, dogrusal_ozellik_seti)

# Hedef değişken (0/1 kodlanmış)
y_train = (df_train_split["satisfaction"] == "satisfied").astype(int)
y_val = (df_val["satisfaction"] == "satisfied").astype(int)

print(f"X_train_agac: {X_train_agac.shape}      X_train_dogrusal: {X_train_dogrusal.shape}")
print(f"X_val_agac:   {X_val_agac.shape}       X_val_dogrusal:   {X_val_dogrusal.shape}")
print(f"X_test_agac:  {X_test_agac.shape}       X_test_dogrusal:  {X_test_dogrusal.shape}")
print(f"y_train: {y_train.shape}   y_val: {y_val.shape}")
```

```python
# ============================================================
# Sınıf Dengesizliğinin Son Teyidi
# ============================================================
sinif_oranlari = y_train.value_counts(normalize=True) * 100
oran = sinif_oranlari[1] / sinif_oranlari[0] if sinif_oranlari[0] > sinif_oranlari[1] else sinif_oranlari[0] / sinif_oranlari[1]

print(f"y_train sınıf dağılımı: %{sinif_oranlari[1]:.1f} memnun, %{sinif_oranlari[0]:.1f} memnun değil")
print(f"Denge oranı: {oran:.2f}:1")
print("SMOTE gibi bir dengeleme tekniğine gerek yok — oran 2:1'in altında, hafif dengesizlik.")
```

```python
# ============================================================
# Uçtan Uca Pipeline — Ham Sütunlardan Başlayan Tek Yapı
# Bu Pipeline; eksik değer doldurma, özellik türetme (log, Ortalama
# Hizmet Puani, Toplam Gecikme, Gecikme Farki, Yas Grubu), encoding
# ve ölçeklemeyi TEK bir nesnede birleştirir. Böylece:
#   - 10 modeli aynı Pipeline'ın sonuna takıp adil şekilde karşılaştırabiliriz
#   - Aynı Pipeline, ileride Flask sitesinde ham kullanıcı girdisini
#     doğrudan alıp tahmine çevirmek için tekrar kullanılabilir
# ============================================================
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.linear_model import LogisticRegression

# OzellikMuhendisligi'yi src/feature_engineering.py'den import ediyoruz
# (notebook içinde ayrıca tanımlamıyoruz). Böylece bu Pipeline'ı joblib ile
# kaydettiğimizde, sınıf her zaman "feature_engineering" modülüne bağlı kalır
# -- src/train.py ve src/predict.py da AYNI modülden import ediyor, bu yüzden
# Flask uygulaması (app/app.py) modeli sorunsuz yükleyebiliyor. Sınıf notebook
# içinde (yani __main__ modülünde) tanımlansaydı, joblib.load() başka bir
# script'ten (örn. app.py) çağrıldığında "Can't get attribute
# 'OzellikMuhendisligi' on <module '__main__'>" hatası verirdi.
import sys
from pathlib import Path
sys.path.append(str(Path("..") / "src"))
from feature_engineering import OzellikMuhendisligi


# Pipeline'ın ölçekleyeceği/encode edeceği sütun grupları
sayisal_olcek_cols = (
    ["Age", "Flight Distance", "Departure Delay in Minutes (log)", "Arrival Delay in Minutes (log)"]
    + service_cols + ["Ortalama Hizmet Puani", "Toplam Gecikme", "Gecikme Farki"]
)
ikili_kategorik_cols = ["Gender", "Customer Type", "Type of Travel"]

on_isleyici = ColumnTransformer(transformers=[
    ("sayisal", StandardScaler(), sayisal_olcek_cols),
    ("ikili", OrdinalEncoder(), ikili_kategorik_cols),
    ("class", OrdinalEncoder(categories=[["Eco", "Eco Plus", "Business"]]), ["Class"]),
    ("yas_grubu", OneHotEncoder(handle_unknown="ignore"), ["Yas Grubu"]),
])

# Ham sütunlar — Pipeline'a bunlar girecek, geri kalanını kendisi türetecek
ham_ozellik_cols = (
    ["Gender", "Customer Type", "Type of Travel", "Class", "Age", "Flight Distance"]
    + gecikme_cols + service_cols
)

X_train_ham = df_train_split[ham_ozellik_cols]
X_val_ham = df_val[ham_ozellik_cols]

# Placeholder model ile Pipeline'ın uçtan uca doğru çalıştığını doğruluyoruz
pipeline_temel = Pipeline([
    ("ozellik_muhendisligi", OzellikMuhendisligi()),
    ("on_isleme", on_isleyici),
    ("model", LogisticRegression(max_iter=1000)),
])

pipeline_temel.fit(X_train_ham, y_train)
dogrulama_skoru = pipeline_temel.score(X_val_ham, y_val)
print(f"Pipeline doğrulama doğruluğu (Logistic Regression, kontrol amaçlı): {dogrulama_skoru:.4f}")
```

```python
# ============================================================
# 10 Model Karşılaştırması — Aynı Pipeline, Farklı Modeller
# Her model, bir önceki cell'de kurduğumuz özellik mühendisliği +
# ön işleme adımlarının (ozellik_muhendisligi + on_isleyici) sonuna
# eklenip, cv_stratejisi (5 katlı stratified CV) ile karşılaştırılıyor.
# ============================================================
from sklearn.model_selection import cross_validate
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier, ExtraTreesClassifier,
    AdaBoostClassifier, GradientBoostingClassifier,
)
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

modeller = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Naive Bayes": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "Extra Trees": ExtraTreesClassifier(n_estimators=200, random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42, algorithm="SAMME"),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "XGBoost": XGBClassifier(random_state=42, eval_metric="logloss"),
    "LightGBM": LGBMClassifier(random_state=42, verbose=-1),
}

sonuclar = []

for isim, model in modeller.items():
    pipeline = Pipeline([
        ("ozellik_muhendisligi", OzellikMuhendisligi()),
        ("on_isleme", on_isleyici),
        ("model", model),
    ])

    cv_sonuc = cross_validate(
        pipeline, X_train_ham, y_train,
        cv=cv_stratejisi,
        scoring=["accuracy", "f1", "roc_auc"],
        n_jobs=-1,
    )

    sonuclar.append({
        "Model": isim,
        "Accuracy": cv_sonuc["test_accuracy"].mean(),
        "F1-Score": cv_sonuc["test_f1"].mean(),
        "ROC-AUC": cv_sonuc["test_roc_auc"].mean(),
        "Süre (sn)": cv_sonuc["fit_time"].sum(),
    })
    print(f"{isim:20s} tamamlandı -> Accuracy: {sonuclar[-1]['Accuracy']:.4f}  F1: {sonuclar[-1]['F1-Score']:.4f}  ROC-AUC: {sonuclar[-1]['ROC-AUC']:.4f}")

model_karsilastirma = pd.DataFrame(sonuclar).sort_values("ROC-AUC", ascending=False).reset_index(drop=True)
model_karsilastirma
```

```python
# ============================================================
# 10 Model İçin Overfitting Gap (Genelleme Riski) ve Test Recall
# Her modeli tam train setinde eğitip, hem train hem test setinde
# 0.5 eşiğiyle değerlendiriyoruz. Overfitting Gap = Train Accuracy -
# Test Accuracy (küçük olması iyi); Test Recall = modelin test
# setindeki gerçek "memnun" yolcuları ne kadar yakaladığı.
# ============================================================
from sklearn.metrics import accuracy_score, recall_score

X_test_ham = df_test[ham_ozellik_cols]
y_test = (df_test["satisfaction"] == "satisfied").astype(int)

ek_metrik_sonuclari = []

for isim, model in modeller.items():
    pipeline = Pipeline([
        ("ozellik_muhendisligi", OzellikMuhendisligi()),
        ("on_isleme", on_isleyici),
        ("model", model),
    ])
    pipeline.fit(X_train_ham, y_train)

    train_tahmin = pipeline.predict(X_train_ham)
    test_tahmin = pipeline.predict(X_test_ham)

    train_acc = accuracy_score(y_train, train_tahmin)
    test_acc = accuracy_score(y_test, test_tahmin)
    test_recall = recall_score(y_test, test_tahmin)

    ek_metrik_sonuclari.append({
        "Model": isim,
        "Overfitting Gap": train_acc - test_acc,
        "Test Recall": test_recall,
    })
    print(f"{isim:20s} tamamlandı -> Overfitting Gap: {ek_metrik_sonuclari[-1]['Overfitting Gap']:.4f}  Test Recall: {test_recall:.4f}")

ek_metrik_df = pd.DataFrame(ek_metrik_sonuclari)
ek_metrik_df
```

    <li><b>Ayarladığımız parametreler</b>: <code>num_leaves</code> ve <code>max_depth</code> (modelin karmaşıklığı/aşırı öğrenme riski), <code>learning_rate</code> ve <code>n_estimators</code> (birlikte çalışırlar — düşük öğrenme hızı + çok ağaç, genelde daha iyi genelleme sağlar), <code>min_child_samples</code> (bir yaprakta olması gereken minimum örnek — aşırı öğrenmeyi frenler), <code>subsample</code> ve <code>colsample_bytree</code> (her ağaç için verinin/özelliklerin bir kısmını rastgele kullanma — yine aşırı öğrenmeye karşı).</li>
    <li><b>Skorlama metriği</b>: <code>roc_auc</code> seçtik, çünkü threshold'dan bağımsız ve modelin olasılıkları ne kadar iyi sıraladığını ölçüyor — 10 model karşılaştırmasında da aynı metriğe göre LightGBM'i seçmiştik, tutarlılık için burada da aynısını kullanıyoruz.</li>
    <li><b>Doğrulama</b>: yine `cv_stratejisi` (5 katlı stratified CV) kullanıyoruz, aynı `X_train_ham`/`y_train` üzerinde — böylece hiperparametre ayarı sırasında da validation/test setine hiç dokunmamış oluyoruz.</li>
  </ul>

  <div class="afn-head">🧭 Karar — Threshold'u Nasıl Seçiyoruz</div>
  <ul>
    <li>Modelimiz her yolcu için 0-1 arası bir <b>olasılık</b> üretiyor; bunu "memnun / memnun değil" etiketine çevirmek için bir eşik gerekiyor. Varsayılan 0.5'in bizim için optimal olduğuna dair hiçbir garanti yok.</li>
    <li>Elimizde havayolu şirketinden gelen "yanlış pozitifin maliyeti şu, yanlış negatifin maliyeti bu" gibi somut bir iş kuralı olmadığı için, en dengeli yaklaşım olarak <b>F1-score'u maksimize eden eşiği</b> seçiyoruz — yani precision ve recall arasında eşit ağırlıklı bir denge kuruyoruz.</li>
    <li>Bunu <b>sadece validation seti</b> (`X_val_ham`/`y_val`) üzerinde yapıyoruz — test setine bu aşamada da dokunmuyoruz, çünkü threshold seçimi de bir tür "öğrenme" ve test setini sadece en sonda, hiç dokunulmamış haliyle final rapor için kullanacağız.</li>
    <li><code>precision_recall_curve</code> ile her olası eşik için precision/recall hesaplayıp, F1'i en yükseğe çıkaran noktayı seçeceğiz; bunu 0.5'teki performansla karşılaştırıp aradaki farkı yorumlayacağız.</li>
  </ul>
</div>


```python
# ============================================================
# Eşik (Threshold) Ayarı — F1-Score'u Maksimize Eden Eşiği Bulma
# ============================================================
from sklearn.metrics import precision_recall_curve, f1_score

y_val_olasilik = lgbm_en_iyi_pipeline.predict_proba(X_val_ham)[:, 1]

precision, recall, esikler = precision_recall_curve(y_val, y_val_olasilik)

# precision/recall dizileri esikler'den bir eleman fazla (son nokta esiksiz), o yuzden hizalıyoruz
f1_skorlari = (2 * precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1] + 1e-9)
en_iyi_index = f1_skorlari.argmax()
en_iyi_esik = esikler[en_iyi_index]

# Karşılaştırma için 0.5 eşiğindeki F1
tahmin_05 = (y_val_olasilik >= 0.5).astype(int)
f1_05 = f1_score(y_val, tahmin_05)

# En iyi eşikteki F1
tahmin_en_iyi = (y_val_olasilik >= en_iyi_esik).astype(int)
f1_en_iyi = f1_score(y_val, tahmin_en_iyi)

print(f"Varsayılan eşik (0.5)   -> F1: {f1_05:.4f}")
print(f"En iyi eşik ({en_iyi_esik:.4f}) -> F1: {f1_en_iyi:.4f}")
print(f"Precision: {precision[en_iyi_index]:.4f}   Recall: {recall[en_iyi_index]:.4f}")
```

```python
# ============================================================
# Final Test Değerlendirmesi — Hiç Dokunulmamış Test Seti
# ============================================================
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
)

SECILEN_ESIK = en_iyi_esik

X_test_ham = df_test[ham_ozellik_cols]
y_test = (df_test["satisfaction"] == "satisfied").astype(int)


def metrikleri_hesapla(pipeline, X, y, esik):
    olasilik = pipeline.predict_proba(X)[:, 1]
    tahmin = (olasilik >= esik).astype(int)
    return {
        "Accuracy": accuracy_score(y, tahmin),
        "Precision": precision_score(y, tahmin),
        "Recall": recall_score(y, tahmin),
        "F1-Score": f1_score(y, tahmin),
        "ROC-AUC": roc_auc_score(y, olasilik),
    }, tahmin


train_metrikleri, _ = metrikleri_hesapla(lgbm_en_iyi_pipeline, X_train_ham, y_train, SECILEN_ESIK)
test_metrikleri, test_tahmin = metrikleri_hesapla(lgbm_en_iyi_pipeline, X_test_ham, y_test, SECILEN_ESIK)

karsilastirma_df = pd.DataFrame({"Train": train_metrikleri, "Test": test_metrikleri})
karsilastirma_df["Fark"] = karsilastirma_df["Train"] - karsilastirma_df["Test"]

print(f"Kullanılan eşik: {SECILEN_ESIK:.4f}\n")
display(karsilastirma_df.round(4))

print("\nTest Seti Sınıflandırma Raporu:")
print(classification_report(y_test, test_tahmin, target_names=["memnun değil / nötr", "memnun"]))

test_conf_matrix = confusion_matrix(y_test, test_tahmin)
print("Karışıklık Matrisi (Test):")
print(test_conf_matrix)
```

```python
# ============================================================
# 5 Fold'un Ayrı Ayrı Sonuçları — Final Model (Ayarlanmış LightGBM)
# ============================================================
from sklearn.base import clone

cv_pipeline = clone(lgbm_en_iyi_pipeline)

cv_sonuclari_detay = cross_validate(
    cv_pipeline, X_train_ham, y_train,
    cv=cv_stratejisi,
    scoring=["f1", "precision", "recall", "roc_auc", "accuracy"],
    n_jobs=-1,
)

cv_detay_df = pd.DataFrame({
    "Fold": [f"Fold {i + 1}" for i in range(cv_stratejisi.get_n_splits())],
    "F1-Score": cv_sonuclari_detay["test_f1"],
    "Precision": cv_sonuclari_detay["test_precision"],
    "Recall": cv_sonuclari_detay["test_recall"],
    "ROC-AUC": cv_sonuclari_detay["test_roc_auc"],
    "Accuracy": cv_sonuclari_detay["test_accuracy"],
})

metrik_kolonlari = ["F1-Score", "Precision", "Recall", "ROC-AUC", "Accuracy"]
ortalama_satiri = pd.DataFrame([{"Fold": "Ortalama", **cv_detay_df[metrik_kolonlari].mean().to_dict()}])
std_satiri = pd.DataFrame([{"Fold": "Std Dev", **cv_detay_df[metrik_kolonlari].std().to_dict()}])

cv_detay_tablo = pd.concat([cv_detay_df, ortalama_satiri, std_satiri], ignore_index=True)

print("📊 CROSS-VALIDATION SONUÇLARI:")
display(cv_detay_tablo.round(4))
```

  <ul>
    <li>Şimdiye kadar modelin <b>ne kadar iyi</b> tahmin ettiğini ölçtük (Accuracy, F1, ROC-AUC). SHAP bize farklı bir soruyu cevaplıyor: modelin tahminlerini verirken <b>hangi özelliklere ne kadar önem verdiğini</b>.</li>
    <li>LightGBM bir ağaç tabanlı model olduğu için, SHAP'ın <code>TreeExplainer</code>'ını kullanıyoruz — bu, ağaç yapısını doğrudan okuyarak çok hızlı ve kesin (yaklaşıksız) SHAP değerleri hesaplıyor.</li>
    <li>Hesaplama maliyetini makul tutmak için validation setinden rastgele 2.000 yolculuk bir örneklem üzerinde çalışıyoruz — bu, güvenilir bir genel görünüm için yeterli.</li>
    <li>Pipeline'ın ön işleme adımlarını (`ozellik_muhendisligi` + `on_isleme`) ayrı çalıştırıp, SHAP'a modelin gerçekten gördüğü <b>dönüştürülmüş</b> özellik matrisini veriyoruz — aksi halde SHAP, ham sütunlar yerine modelin iç temsilini açıklayamazdı.</li>
  </ul>

```python
# ============================================================
# SHAP Global Özellik Önemi — Final Model (LightGBM)
# ============================================================
np.random.seed(42)
shap_ornek_ham = X_val_ham.sample(n=min(2000, len(X_val_ham)), random_state=42)

on_isleme_adimi = lgbm_en_iyi_pipeline.named_steps["ozellik_muhendisligi"]
donusum_adimi = lgbm_en_iyi_pipeline.named_steps["on_isleme"]
lgbm_model_adimi = lgbm_en_iyi_pipeline.named_steps["model"]

X_ozellik_muhendisligi = on_isleme_adimi.transform(shap_ornek_ham)
X_donusturulmus = donusum_adimi.transform(X_ozellik_muhendisligi)
ozellik_isimleri_ham = donusum_adimi.get_feature_names_out()
ozellik_isimleri = [isim.split("__", 1)[-1] for isim in ozellik_isimleri_ham]

explainer = shap.TreeExplainer(lgbm_model_adimi)
shap_degerleri_ham = explainer.shap_values(X_donusturulmus)

# Bazı SHAP/sürüm kombinasyonlarında binary sınıflandırma icin liste donebiliyor
if isinstance(shap_degerleri_ham, list):
    shap_degerleri = shap_degerleri_ham[1]
else:
    shap_degerleri = shap_degerleri_ham

ortalama_mutlak_shap = np.abs(shap_degerleri).mean(axis=0)

shap_onem_df = pd.DataFrame({
    "Özellik": ozellik_isimleri,
    "Ortalama Mutlak SHAP Etkisi": ortalama_mutlak_shap,
}).sort_values("Ortalama Mutlak SHAP Etkisi", ascending=False).reset_index(drop=True)

display(shap_onem_df.head(15))
```
